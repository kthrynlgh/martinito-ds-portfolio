## **Overview**

A **feedforward neural network** is an artificial neural network in which information moves in one direction: from the input layer, through the hidden layer(s), and toward the output layer.

Its basic components are:

- **Input layer:** receives the input features.
- **Hidden layer:** computes weighted sums, adds biases, and applies an activation function.
- **Output layer:** produces the final prediction.

A forward pass follows this sequence:

```mermaid
graph LR
    Input["Input layer"] --> Hidden["Hidden layer"]
    Hidden --> Output["Output layer"]
    Output --> Prediction["Prediction"]
```
<br>

For each neuron, the weighted sum is:

$$
Z = \sum_i w_i x_i + \theta
$$

where $w_i$ is a weight, $x_i$ is an input, and $\theta$ is the bias.
<br>

An activation function is then applied to $Z$. Backpropagation may be used after the forward pass during training, but it is not part of the forward-pass calculation itself.

### **Activation Function: ReLU**

The laboratory task specifies:

$$
f(Z_n)=\max(0,Z_n)
$$
<br>

This is the **Rectified Linear Unit (ReLU)** activation function:

$$
\operatorname{ReLU}(Z)=
\begin{cases}
0, & Z<0\\
Z, & Z\geq 0
\end{cases}
$$

### **Setup**

Input:

x=\begin{bmatrix}1\\0\\1\end{bmatrix}

## **Implementation**

### **Import Libraries**

````python
import numpy as np
````

### **Define the Inputs, Weights, and Biases**

````python
x = np.array([1, 0, 1], dtype=float)
y = 1.0

W_hidden = np.array([
    [0.2, -0.3],
    [0.4,  0.1],
    [-0.5, 0.2]
], dtype=float)

W_output = np.array([-0.3, -0.2], dtype=float)

theta_hidden = np.array([-0.4, 0.2], dtype=float)
theta_output = 0.1
````

### **Define the ReLU Function**

The ReLU function returns the maximum between zero and the input value.

````python
def relu(z):
    return np.maximum(0, z)
````

### **Calculate the Hidden Layer**

***Hidden layer 1***

$$
Z_1=(0.2)(1)+(0.4)(0)+(-0.5)(1)-0.4=-0.7
$$

$$
H_1=\operatorname{ReLU}(-0.7)=0
$$
<br>

***Hidden layer 2***

$$
Z_2=(-0.3)(1)+(0.1)(0)+(0.2)(1)+0.2=0.1
$$

$$
H_2=\operatorname{ReLU}(0.1)=0.1
$$

````python
Z_hidden = x @ W_hidden + theta_hidden
H = relu(Z_hidden)

print("Hidden-layer weighted sums:", Z_hidden)
print("Hidden-layer outputs:", H)
````

### **Calculate the Output Layer**

The output neuron receives the hidden-layer outputs:

$$
Z_3=(-0.3)(H_1)+(-0.2)(H_2)+0.1
$$
<br>

Substituting $H_1=0$ and $H_2=0.1$:

$$
Z_3=(-0.3)(0)+(-0.2)(0.1)+0.1=0.08
$$
<br>

Applying ReLU:

$$
\hat{y}=\operatorname{ReLU}(0.08)=0.08
$$

````python
Z_output = H @ W_output + theta_output
y_hat = relu(Z_output)

print("Output weighted sum:", Z_output)
print("Predicted output:", y_hat)
````

### **Calculate the Error**

Because the task does not specify an error formula, we will use the single-output squared error:

$$
E=\frac{1}{2}(y-\hat{y})^2
$$
<br>

Using $y=1$ and $\hat{y}=0.08$:

$$
E=\frac{1}{2}(1-0.08)^2=0.4232
$$

````python
error = 0.5 * (y - y_hat) ** 2

print(f"Target output: {y:.2f}")
print(f"Predicted output: {y_hat:.2f}")
print(f"Squared error: {error:.4f}")
````

**Interpretation**

During the forward pass, the first hidden layer yields an output of $0$ because its pre-activation sum is negative ($Z_1 = -0.7$), causing the ReLU activation function to suppress it. The second hidden layer produces an output of $0.1$ ($H_2 = 0.1$), allowing positive signals to flow into the output layer to generate the final prediction of $\hat{y} = 0.08$.

**Conclusion**

The network successfully completed its initial forward pass, mapping the input through linear and non-linear transformations to produce a squared error of $E = 0.4232$ against a target of $1.00$. Because this error is substantial, the network will require subsequent training iterations (via backpropagation) to adjust its weights and improve future predictions.
